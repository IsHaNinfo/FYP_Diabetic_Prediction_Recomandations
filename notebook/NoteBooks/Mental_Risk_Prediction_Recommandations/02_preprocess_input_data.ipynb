```python
import pandas as pd

# Step 1: Load CSV
df = pd.read_csv('../../data/MentalParameter.csv')

# Step 2: Clean column names
df.columns = df.columns.str.replace(r'\n', '', regex=True).str.strip()

# Step 3: Define mapping dictionary
value_map = {
    'Not at all': 1,
    'Never': 1,
    'Rarely': 2,
    'Sometimes': 3,
    'Several days': 3,
    'More than half the days': 4,
    'Often': 5,
    'Nearly every day': 5,
    'Always (6+ times)': 5,
    'Often  (4-5 times)': 4,
    'Sometimes (2-3 times)': 3,
    'Rarely  (0-1 times)': 2
}

# Step 4: List of columns to map
categorical_columns = [
    'Stress Frequency and Intensity (Weight: 15%)',
    'Emotional Regulation',
    'Physical Symptoms of Stress',
    'Cognitive Stress (Negative Self-Thoughts)',
    'Behavioral Response',
    'Concentration and Productivity (Weight: 10%)',
    'Suicidal Thoughts (Optional, Use with Caution)'
]

# Step 5: Apply mapping
for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].map(value_map)

# Step 6: Normalize Free Time (ranking-based bins: 1 = low, 5 = high)
if 'Available Time' in df.columns:
    ranked = df['Available Time'].rank(method='min') 
    df['Available Time Normalized'] = pd.cut(
        ranked,
        bins=5,
        labels=[1, 2, 3, 4, 5]
    ).astype(int)

# Step 7: Save cleaned data
df.to_csv('../../data/MentalParameter_clean.csv', index=False)
```
