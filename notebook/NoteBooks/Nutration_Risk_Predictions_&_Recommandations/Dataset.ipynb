## Merge two nutritons data set

```python
import pandas as pd
```

```python
sri_lankan_nutrition_foods_1 = pd.read_csv('../../data/RecommandationDatasets/Sri_Lankan_Nutritions.csv') 
sri_lankan_nutrition_foods_2 = pd.read_csv('../../data/RecommandationDatasets/SrilankanCommonFoods.csv')
sri_lankan_nutrition_foods_3 = pd.read_csv('../../data/RecommandationDatasets/sri_lankan_food_types.csv')
# Load the food_items dataset
```

```python
sri_lankan_nutrition_foods_1.head()
```

```python
sri_lankan_nutrition_foods_1.drop(columns=["FCT source","drfoodlistcode"], inplace=True)
```

```python

# Clean all column names
sri_lankan_nutrition_foods_1.columns = sri_lankan_nutrition_foods_1.columns.str.strip()                     # Removes leading/trailing spaces and \n
sri_lankan_nutrition_foods_1.columns = sri_lankan_nutrition_foods_1.columns.str.replace('\n', '', regex=True)  # Removes newlines
sri_lankan_nutrition_foods_1.columns = sri_lankan_nutrition_foods_1.columns.str.replace(' ', '_')           # Optional: Replace spaces with underscores

sri_lankan_nutrition_foods_1.rename(columns={
    'drfoodlistdesc': 'Food',
    'Energy_(kcal)':'Calories_(kcal)',
    'Total_fats_(g)': 'Fats_(g)',
    'Total_carbohydrates_(g)': 'Carbohydrates_(g)',
}, inplace=True)
sri_lankan_nutrition_foods_1.head()
```

```python

# Clean all column names
sri_lankan_nutrition_foods_2.columns = sri_lankan_nutrition_foods_2.columns.str.strip()                     # Removes leading/trailing spaces and \n
sri_lankan_nutrition_foods_2.columns = sri_lankan_nutrition_foods_2.columns.str.replace('\n', '', regex=True)  # Removes newlines
sri_lankan_nutrition_foods_2.columns = sri_lankan_nutrition_foods_2.columns.str.replace(' ', '_')           # Optional: Replace spaces with underscores

sri_lankan_nutrition_foods_2.rename(columns={
    'drfoodlistdesc': 'Food',
    'Calories (kcal)':'Calories_(kcal)',
    'Fat (g)': 'Fats_(g)',
    'Carbohydrate (g)': 'Carbohydrates_(g)',
}, inplace=True)
sri_lankan_nutrition_foods_2.head()
```

```python
sri_lankan_nutrition_foods_3.head()
```

```python
sri_lankan_nutrition_foods_3.drop(columns=["Food ID","Region","Price (LKR)"], inplace=True)
```

```python
sri_lankan_nutrition_foods_3.columns = sri_lankan_nutrition_foods_3.columns.str.strip()                     # Removes leading/trailing spaces and \n
sri_lankan_nutrition_foods_3.columns = sri_lankan_nutrition_foods_3.columns.str.replace('\n', '', regex=True)  # Removes newlines
sri_lankan_nutrition_foods_3.columns = sri_lankan_nutrition_foods_3.columns.str.replace(' ', '_')           # Optional: Replace spaces with underscores

sri_lankan_nutrition_foods_3.rename(columns={
    'Food_Name': 'Food',
    'Calories': 'Calories_(kcal)',
}, inplace=True)
sri_lankan_nutrition_foods_3.head()
```

```python
sri_lankan_nutrition_foods_2.head()
```

### Merge Datasets

```python
# Ensure 'Food' column is clean and consistent in all DataFrames
for df in [sri_lankan_nutrition_foods_1, sri_lankan_nutrition_foods_2, sri_lankan_nutrition_foods_3]:
    df['Food'] = df['Food'].astype(str).str.strip().str.upper()

# Merge all three DataFrames on 'Food'
merged = pd.merge(
    sri_lankan_nutrition_foods_2, 
    sri_lankan_nutrition_foods_3, 
    on='Food', 
    how='outer', 
)
merged_all = pd.merge(
    merged, 
    sri_lankan_nutrition_foods_1, 
    on='Food', 
    how='outer', 
    suffixes=('', '_nutritions')
)
```

```python
merged_all.head()
```

```python
# Combine main nutrition columns
for col in ['Calories_(kcal)', 'Carbohydrate_(g)', 'Protein_(g)', 'Fat_(g)']:
    x_col = f"{col}_x"
    y_col = f"{col}_y"
    if x_col in merged_all.columns and y_col in merged_all.columns:
        merged_all[col] = merged_all[x_col].combine_first(merged_all[y_col])
        merged_all = merged_all.drop(columns=[x_col, y_col])
```

```python
merged_all.head()
```

```python
for col, unit in [
    ('Carbohydrate_(g)', 'g'),
    ('Quantity','g'),
    ('Protein_(g)', 'g'),
    ('Fat_(g)', 'g'),
    ('Calories_(kcal)', 'kcal')
]:
    if col in merged_all.columns:
        merged_all[col] = (
            merged_all[col]
            .astype(str)
            .str.replace(unit, '', regex=False)
            .str.strip()
        )
        merged_all[col] = pd.to_numeric(merged_all[col], errors='coerce')
```

```python
merged_all.head()
```

```python
if 'Recipe/Single_food_item' in merged_all.columns:
    merged_all['Recipe/Single_food_item'] = merged_all['Recipe/Single_food_item'].fillna('S')
```
